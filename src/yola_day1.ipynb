```python
from ultralytics import YOLO 
```

```python
import os 
from pathlib import Path

image_dir = Path(r"C:\Users\Honor\Desktop\CV_ete\YOLA\YOLA_beginner2026")
image = os.listdir(image_dir / "input" )
model = YOLO("models/yolov8n.pt")
no_lable = []
for i, img in enumerate(image):
    results = model(image_dir / "input" / img )
    results[0].save(f"{image_dir}/outputs/result_{img}")
    if len(results[0]) == 0 :
        no_lable.append(i)
    else: no_lable.append(-1)
```

```python
import matplotlib.pyplot as plt 
from PIL import Image
import numpy as np

img_array = [np.array(Image.open(f"{image_dir}/outputs/result_{img}")) for img in image]

figs, axes = plt.subplots(2,5 , figsize= (15, 9))

for i, axs in enumerate(axes.flatten()):
    axs.imshow(img_array[i])
    if no_lable[i] == -1 :
        axs.set_title((image[i]).split('.')[0])
    else : axs.set_title("no_lable")
    axs.axis('off')

plt.tight_layout()
plt.show()
```
